**Задание 1**

Загрузите из CSV-файла исходные данные о покупателях с веб-портала (до 15 строк). Проанализируйте возраст:

если больше 25 лет, то выводить «Старше 25»;
если меньше или равно, то выводить «Младше или равен 25».

Сделать через применение функции .apply pandas, изменить данные столбца DataFrame по условию.
Вывод задания 1 должен быть равен:

0 Старше 25
1 Старше 25
2 Старше 25
3 Старше 25
4 Младше или равен 25
5 Старше 25
6 Старше 25
7 Старше 25
8 Старше 25
9 Старше 25
10 Старше 25
11 Старше 25
12 Младше или равен 25
13 Старше 25
14 Старше 25

```python
# Загрузка библитотеки
import pandas as pd

# Загружаем данные
def load_data_client():
    """Загружает первые 15 строк из CSV файла"""
    filename = "web_clients_correct_доп материал к тема 6, 7.csv"
    data = pd.read_csv(filename, nrows=15)
    return data

# Классификации возраста
def classify_age(age):
    """Определяет категорию возраста"""
    if age > 25:
        return "Старше 25"
    else:
        return "Младше или равен 25"

# Основная функция для задания 1
def main_client():
    # Загружаем данные
    df = load_data_client()

    # Применяем функцию classify_age к столбцу age через .apply()
    age_categories = df['age'].apply(classify_age)

    # Выводим результат
    print("Результат Задания 1:")
    for index, category in age_categories.items():
        print(f"{index} {category}")
        # Запуск задания 1
if __name__ == "__main__":
    main_client()
```

**Задание 2**

Загрузить данные из формата CSV о покупателях с веб-портала (до 15 строк). Отобрать DataFrame по нескольким условиям поля «sex» и «age», пол — мужской, возраст — больше 30.


```python
# Загрузка библиотеки
import pandas as pd

# Загружаем 15 строк из CSV по условиям задания
df = pd.read_csv("web_clients_correct_доп материал к тема 6, 7.csv", nrows=15)

# Фильтруем: мужчины и старше 30 лет
result = df[(df['sex'] == 'male') & (df['age'] > 30)]

# Выводим таблицу
print("Результат Задания 2:")
print(result.to_string())
```
