# Задание 1
 Пользователь по команде «p» может узнать владельца документа по его номеру.

Пример работы программы:
Введите команду:
p

Введите номер документа:
10006

Результат:
Владелец документа: Аристарх Павлов

# Решение
С помощью создания функций, поиск for if. Цикл while для работы верных команд

```python
# Исходные данные
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}

def find_owner():

    doc_number = input("Введите номер документа: ")

    # Поиск через цикл
    for doc in documents:
        if doc['number'] == doc_number:
            print(f"Владелец документа: {doc['name']}")
            return

    # Если не нашли
    print("Документ с таким номером не найден")

# Выводим программу
print("Программа учета документов")
print("Команды: p - найти владельца, q - выход")

while True:
    command = input("Введите команду: ")

    if command == "q":
        print("Выход из программы")
        break
    elif command == "p":
        find_owner()
    else:
        print("Неизвестная команда")
```

# Задание 2 (необязательное)
 Секретарь по команде «s» может по номеру документа узнать, на какой полке документ хранится.

Пример работы программы:

Введите команду:
s

Введите номер документа:
10006

Результат:
Документ хранится на полке: 2

```python
# Исходные данные
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}

def find_owner():
    doc_number = input("Введите номер документа: ")

    # Поиск через цикл
    for doc in documents:
        if doc['number'] == doc_number:
            print(f"Владелец документа: {doc['name']}")
            return

    # Если не нашли
    print("Документ с таким номером не найден")

def find_shelf():
    doc_number = input("Введите номер документа: ")

    # Поиск полки через цикл
    for shelf, docs in directories.items():
        if doc_number in docs:
            print(f"Документ хранится на полке: {shelf}")
            return

    # Если не нашли
    print("Документ с таким номером не найден")

# Выводим программу
print("Программа учета документов")
print("Команды: p - найти владельца, s - найти полку, q - выход")

while True:
    command = input("Введите команду: ")

    if command == "q":
        print("Выход из программы")
        break
    elif command == "p":
        find_owner()
    elif command == "s":
        find_shelf()
    else:
        print("Неизвестная команда")
```
