```python
import pandas as pd
import plotly as pl

def load_data(filename, n_rows=15):
    """Загружаем данные из CSV файла"""
    try:
        df = pd.read_csv(filename)
        return df.head(n_rows)
    except FileNotFoundError:
        print(f"Файл {filename} не найден!")
        return None

def create_scatter_plot(df):
    """Создаем scatter plot: зависимости возраста от расходов"""

    fig = px.scatter(
        df,
        x='age',
        y='bill',
        title='Зависимость расходов от возраста клиентов',
        labels={'age': 'Возраст', 'bill': 'Расходы'},
        hover_data=['name', 'sex']
    )

    # Настраиваем внешний вид
    fig.update_layout(
        xaxis_title='Возраст',
        yaxis_title='Расходы',
        showlegend=False,
        width=800,
        height=600,
        font=dict(size=12)
    )

    # Настраиваем оси x и y
    fig.update_xaxes(range=[20, 85], dtick=10)
    fig.update_yaxes(range=[0, 1200], dtick=200)

    return fig

def main():
    """Основная функция программы"""

    # Загружаем данные
    df = load_data("web_clients_correct_доп материал к тема 6, 7.csv", 15)

    if df is None:
        print("Не удалось загрузить данные!")
        return

    # Создаем график
    fig = create_scatter_plot(df)

    # Показываем график
    fig.show()

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Загружаем данные
df = pd.read_csv("web_clients_correct_доп материал к тема 6, 7.csv", nrows=15)

# Создаем график с линией регрессии
plt.figure(figsize=(10, 6))
sns.regplot(
        data=df,
        x='age',
        y='bill',
        scatter_kws={
            'alpha': 0.9,
            'color': 'blue',
            's': 30  # размер точек
        },
        line_kws={
            'color': 'red',
            'linewidth': 2,
            'label': 'Линия регрессии'
        }
    )

# Добавляем заголовок и подписи
plt.title('Зависимость расходов от возраста клиентов')
plt.xlabel('Возраст')
plt.ylabel('Расходы')

# Показываем график
plt.show()
```
