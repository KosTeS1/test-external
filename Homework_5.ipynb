# Задание
Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Используйте для работы файл purchase_log.txt.
Если покупка была, сам файл visit_log.csv изменять не надо.
Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.

Учтите условие для данных:

содержимое visit_log.csv — не помещается в оперативную память компьютера, используйте только построчную обработку этого файла.

Примеры работы программы при выводе первых трёх строк файла funnel.csv:
user_id,source,category
1840e0b9d4,other,Продукты
4e4f90fcfb,context,Электроника

# Решение:
Думаю, что стоит решить задачу таким способом:

*   Файл purchase_log.txt - сделать словарём
*   visit_log.csv - провести построчную обработка согласно заданию
*   funnel.csv - запись результатов покупки с указанием категории, формат user_id,source,category

```python
# Создаем словарь покупок
purchase_dict = {}

with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    next(f)

    for line in f:
        try:
            data = json.loads(line.strip())
            user_id = data.get('user_id')
            category = data.get('category')

            if user_id and category and user_id != 'user_id':
                purchase_dict[user_id] = category
        except:
            continue

print(f"Загружено {len(purchase_dict)} покупок")

# Обрабатываем visit_log.csv построчно
with open('visit_log.csv', 'r', encoding='utf-8') as visit_file, \
     open('funnel.csv', 'w', encoding='utf-8') as output_file:

    # Записываем заголовок
    output_file.write('user_id,source,category\n')

    # Пропускаем заголовок visit_log.csv
    header = visit_file.readline()

    matched_count = 0

    # Читаем файл построчно
    for line in visit_file:
        parts = line.strip().split(',')

        if len(parts) >= 2:
            user_id = parts[0]
            source = parts[1]

            # Если была покупка - записываем
            if user_id in purchase_dict:
                category = purchase_dict[user_id]
                output_file.write(f'{user_id},{source},{category}\n') #\n перенос строк
                matched_count += 1

print(f"Создан файл funnel.csv с {matched_count} записями")
```
